## Search


A general search pseudocode is as follows:

```
Here's a useful pseudocode for search, in general:

    Start with a frontier that contains the initial state
    Start with an empty explored set
    Repeat:
        If the frontier is empty, then no solution
        Remove a node from the frontier
        If node contains goal state, return the solution
        Add the node to the explored set
        Expand node, add resulting nodes to the frontier if they aren't already in the frontier or the explored set
```

Now, this is the general idea. How we deal with the frontier (implementation, queue, priority queue, stack, list, dictionary, and so on) and some of the conditions is going to differentiate whether this leads to breadth-first, depth-first, or A$^*$.


In class we talked about alphabetical order, but here we're dealing with locations in a list of lists. Let moves dictate the order. To arrive at the expected output, do not amend the order in which actions are arranged in moves (Up, Right, Down, Left).

````python
from copy import deepcopy
import math
import random
random.seed(42)  # keep random results the same each time
````

````python
from collections import deque
import heapq

def _open(world, pos):
    """Check if this position can be used."""
    r, c = pos
    return 0 <= r < len(world) and 0 <= c < len(world[0]) and world[r][c] != '⬛'

def _path(world, parent, start, goal):
    """Mark the path from the goal back to the start."""
    p = parent.get(goal)
    while p is not None and p != start:
        if world[p[0]][p[1]] == '🟨':
            world[p[0]][p[1]] = '🟩'
        p = parent[p]
    return world

def _bfs(world, start, goal, moves):
    """BFS: explore the oldest position first."""
    out, q = deepcopy(world), deque([start])
    parent, seen = {start: None}, {start}
    while q:
        cur = q.popleft()
        if cur == goal: return _path(out, parent, start, goal)
        for dr, dc in moves:
            nxt = (cur[0] + dr, cur[1] + dc)
            if _open(out, nxt) and nxt not in seen:
                seen.add(nxt); parent[nxt] = cur; q.append(nxt)
                if nxt not in (start, goal) and out[nxt[0]][nxt[1]] == '⬜': out[nxt[0]][nxt[1]] = '🟨'
    return out

def _dfs(world, start, goal, moves):
    """DFS: explore the newest position first."""
    out, stack = deepcopy(world), [start]
    parent, seen = {start: None}, {start}
    while stack:
        cur = stack.pop()
        if cur == goal: return _path(out, parent, start, goal)
        for dr, dc in reversed(moves):
            nxt = (cur[0] + dr, cur[1] + dc)
            if _open(out, nxt) and nxt not in seen:
                seen.add(nxt); parent[nxt] = cur; stack.append(nxt)
                if nxt not in (start, goal) and out[nxt[0]][nxt[1]] == '⬜': out[nxt[0]][nxt[1]] = '🟨'
    return out

def _astar(world, start, goal, moves):
    """Perform A-star search with unit costs and straight-line heuristic."""
    out, heap, count = deepcopy(world), [], 0
    g, parent, closed = {start: 0}, {start: None}, set()
    h = lambda p: math.hypot(p[0] - goal[0], p[1] - goal[1])
    heapq.heappush(heap, (h(start), 0, count, start)); count += 1
    while heap:
        _, gc, _, cur = heapq.heappop(heap)
        if cur in closed: continue
        closed.add(cur)
        if cur == goal: return _path(out, parent, start, goal)
        for dr, dc in moves:
            nxt, ng = (cur[0] + dr, cur[1] + dc), gc + 1
            if _open(out, nxt) and nxt not in closed and ng < g.get(nxt, math.inf):
                g[nxt], parent[nxt] = ng, cur
                heapq.heappush(heap, (ng + h(nxt), ng, count, nxt)); count += 1
                if nxt not in (start, goal) and out[nxt[0]][nxt[1]] == '⬜': out[nxt[0]][nxt[1]] = '🟨'
    return out
````

## Adversarial Search


The MINIMAX pseudocode is as follows:

```
MINIMAX(state) returns action
    return argmax action in Actions MIN-VALUE(RESULT(state,action))
    

MAX-VALUE(state) returns value
if TERMINAL(state) return UTILITY(state)
v <- -inf
for each action in Actions(state)
    v <- MAX(v, MIN-VALUE(RESULT(state, action))
return v


MIN-VALUE(state) returns value
if TERMINAL(state) return UTILITY(state)
v <- +inf
for each action in Actions(state)
    v <- MIN(v, MAX-VALUE(RESULT(state, action))
return v
```


## Notes.

* MINIMAX is used to model the opponent as a rational player that will also try it's best to win. When moves are simulated in the tree, that's just what they are, simulations. The actual board only changes once MINIMAX returns an action and that action is taken.
* You will probably need to maintain a deep copy of the world.
* Use a depth parameter to control the number of times the algorithm ping pongs between the calls to the min_value and max_value functions. A depth of 2 would mean P1_MAX's (caller of MINIMAX) moves are simulated and then P2_MIN's moves are simulated, and that's it.
* Responses to FAQ (prior semester):
* * There is no "act of zapping" special action taken by '👾' that costs it a move
* * * '🤖' is simply zapped by '👾' if it is anywhere in a 3x3 cell centered at the bad robot
* * Paths can be crossed. It's possible for a robot to occupy a cell previously occupied by the other robot.
* * It's possible for a robot to re-visit a position. When looking at results, remember game over happens when either player wins or we've exhausted the 82 plies (41 turns each by each robot).
* It might be useful to print out each ply or turn for debugging (not final submission)
* The utility formulas used in this project are not perfect, but balance multiple factors: implementation simplicity, consistent outputs, and grading efficiency
* * The '🤖' utility in Problem #1 is trying to encode that farther from '👾' is better and closer to '❤️' better
* * The '👾' utility in Problem #1 is trying to encode that closer to '🤖' is better and only cares about zapping
* * The '👾' utility in Problem #3 represents a slightly smarter '👾' that wants to zap but also thinks being near '❤️' is good for zapping '🤖'

````python
def _zapped(good, bad):
    """Check whether the good robot is in the bad robot's 3x3 range."""
    return abs(good[0] - bad[0]) <= 1 and abs(good[1] - bad[1]) <= 1

def _terminal(good, bad, goal):
    """Return 1 for a good win, -1 for a bad win, or 0 otherwise."""
    if good == goal:
        return 1
    if _zapped(good, bad):
        return -1
    return 0

def dist(a, b):
    """Return Euclidean distance between two grid positions."""
    return math.hypot(a[0] - b[0], a[1] - b[1])

def _legal(world, pos, moves):
    """Return legal destinations from a robot position."""
    return [(pos[0] + dr, pos[1] + dc) for dr, dc in moves
            if _open(world, (pos[0] + dr, pos[1] + dc))]

def _utility(player, good, bad, goal, aware):
    """Calculate the utility for the selected player."""
    if player == "good":
        return dist(bad, good) - dist(good, goal)
    if aware:
        return -dist(bad, goal) - dist(bad, good)
    return -dist(good, bad)

def _mm_value(world, good, bad, goal, moves, root, turn, depth, aware):
    """Evaluate a game state with alternating minimax turns."""
    terminal = _terminal(good, bad, goal)
    if terminal:
        if terminal == 1:
            return math.inf if root == "good" else -math.inf
        return -math.inf if root == "good" else math.inf
    if depth == 0:
        return _utility(root, good, bad, goal, aware)

    pos = good if turn == "good" else bad
    options = _legal(world, pos, moves)
    if not options:
        return _utility(root, good, bad, goal, aware)

    values = []
    for nxt in options:
        ng, nb = (nxt, bad) if turn == "good" else (good, nxt)
        other = "bad" if turn == "good" else "good"
        values.append(_mm_value(world, ng, nb, goal, moves,
                                root, other, depth - 1, aware))

    return max(values) if turn == root else min(values)
````

````python
def _choose_move(world, good, bad, goal, moves, player, strategy, depth):
    """Choose a legal move using random choice or minimax."""
    pos = good if player == "good" else bad
    options = _legal(world, pos, moves)
    if strategy == "RANDOM":
        return random.choice(options)

    aware = strategy == "MINIMAX-AWARE"
    best = options[0]
    best_value = -math.inf

    for nxt in options:
        ng, nb = (nxt, bad) if player == "good" else (good, nxt)
        other = "bad" if player == "good" else "good"
        value = _mm_value(world, ng, nb, goal, moves,
                          player, other, depth - 1, aware)
        if value > best_value:
            best, best_value = nxt, value
    return best
````

## Problem #1


Implement the game: players, actions, scores, keeping track of turns, when the game is over, how the world changes, and what strategy each player uses (random moves or MINIMAX).


Play the game:
* '🤖' will use MINIMAX with a depth of 2 (simulate I move and then my opponent moves -- that's it)
* '🤖' utility function = Euclidean distance to '👾' - Euclidean distance to '❤️'
* '👾' will choose actions at random


When the game is over, print out the world (nicer view) with the following changes:
* Replace a '⬜' with '🟥' to indicate the path of '👾',
* Replace '⬜' with '🟩' to indicate the path of '🤖',
* Places the '🤖' and '👾' icons at their current positions, and
* '❤️' and '⬛' are not replaced.

````python
def run_robot_game(world, good_start, bad_start, goal, moves, good_strategy, good_depth, bad_strategy, bad_depth): # AUTOGRADER: DO NOT MODIFY FUNCTION HEADER
    """Run the robot game and show both paths."""
    out, good, bad = deepcopy(world), good_start, bad_start
    good_path, bad_path = {good}, {bad}
    for ply in range(82):
        player = "good" if ply % 2 == 0 else "bad"
        strategy = good_strategy if player == "good" else bad_strategy
        depth = good_depth if player == "good" else bad_depth
        nxt = _choose_move(out, good, bad, goal, moves, player, strategy, depth)
        if player == "good": good = nxt; good_path.add(good)
        else: bad = nxt; bad_path.add(bad)
        if _terminal(good, bad, goal): break
    for r in range(len(out)):
        for c in range(len(out[0])):
            p = (r, c)
            if p not in (good, bad, goal) and out[r][c] == "⬜":
                out[r][c] = "🟩" if p in good_path else ("🟥" if p in bad_path else "⬜")
    out[good[0]][good[1]], out[bad[0]][bad[1]], out[goal[0]][goal[1]] = "🤖", "👾", "❤️"
    return out
````
